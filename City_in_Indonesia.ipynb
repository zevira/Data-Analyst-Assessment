```python
# Convert all string columns to uppercase, except Funcloc and RegionalCode
df_city_indonesia = df_city_indonesia.apply(
    lambda x: x.str.upper() if x.name not in ['RegionalCode'] and x.dtype == 'object' else x
)

df_assessment = df_assessment.apply(
    lambda x: x.str.upper() if x.name not in ['Funcloc'] and x.dtype == 'object' else x
)
```

```python
# Menampilkan nilai unik pada kolom City
unique_cities = df_city_indonesia['City'].unique()

# Menampilkan hasil
print(unique_cities)
```

```python
# Menghapus kata "Kabupaten" atau "Kota" dari kolom City
df_city_indonesia['City'] = df_city_indonesia['City'].str.replace(r'\b(KABUPATEN|KOTA)\b', '', regex=True).str.strip()

# Menampilkan hasil
print(df_city_indonesia['City'])
```

```python
# Menampilkan nilai unik pada kolom City
unique_cities = df_city_indonesia['City'].unique()

# Menampilkan hasil
print(unique_cities)
```

```python
# Menghapus kata "ADM." dari kolom City
df_city_indonesia['City'] = df_city_indonesia['City'].str.replace('ADM.', '', regex=False).str.strip()
```

```python
df_city_indonesia['City'] = df_city_indonesia['City'].replace({
    'TOJO UNA-UNA': 'TOJO UNA UNA',
    'OGAN KOMERING ULU SELATAN (OKU SELATAN)': 'OKU SELATAN',
    'OGAN KOMERING ULU TIMUR (OKU TIMUR)': 'OKU TIMUR',
    'SIDENRENG RAPPANG (SIDRAP)': 'SIDENRENG RAPPANG',
    'SURAKARTA (SOLO)': 'SURAKARTA',
    'LIMA PULUH': 'LIMA PULUH KOTA',
    'KUTAI KARTANEGARA': 'KUTAI KERTANEGARA',
    'FAKFAK': 'FAK FAK',
    'KEPULAUAN YAPEN (YAPEN WAROPEN)': 'KEPULAUAN YAPEN',
    'LABUHANBATU SELATAN': 'LABUHAN BATU SELATAN',
    'LABUHANBATU UTARA': 'LABUHAN BATU UTARA',
    'LABUHANBATU': 'LABUHAN BATU',
    'POHUWATO': 'PAHUWATO',
    'PANGKAJENE KEPULAUAN': 'PANGKAJENE'
})
```

```python
df_city_indonesia['City'] = df_city_indonesia['City'].replace({
    'KEPULAUAN SIAU TAGULANDANG BIARO (SITARO)': 'KEPULAUAN SIAU TAGULANDANG BIARO'
})
```

```python
# Menampilkan nilai unik pada kolom City
unique_cities = df_city_indonesia['City'].unique()

# Menampilkan hasil
print(unique_cities)
```

```python
# Menampilkan nilai unik pada kolom Alamat4
unique_alamat4 = df_assessment['Alamat4'].unique()

# Menampilkan hasil
print(unique_alamat4)
```

```python
# Menghapus ', KOTA' dan 'KAB.' dari kolom Alamat4
df_assessment['Alamat4'] = (
    df_assessment['Alamat4']
    .str.replace(', KOTA', '', regex=False)
    .str.replace('KAB.', '', regex=False)
    .str.strip()  # Menghapus spasi berlebih di awal/akhir string
)

# Menampilkan hasil
print(df_assessment['Alamat4'])
```

```python
df_assessment['Alamat4'] = df_assessment['Alamat4'].replace({
    'KEPULAUAN SIAU TAGULANDAN': 'KEPULAUAN SIAU TAGULANDANG BIARO'
})
```

```python
df_assessment['Alamat4'] = df_assessment['Alamat4'].replace({
    'LABUHANBATU UTARA': 'LABUHAN BATU UTARA',
    'LABUHANBATU SELATAN': 'LABUHAN BATU SELATAN',
    'PANGKAJENE KEPULAUAN': 'PANGKAJENE',
    'LABUHANBATU': 'LABUHAN BATU'
})
```

```python
df_assessment.loc[df_assessment['Alamat1'] == 'GD.GRAHAPRATAMALT.5,JL.MT HARYON', 'Alamat4'] = 'JAKARTA SELATAN'
```

```python
# Menampilkan nilai unik pada kolom Alamat4 dan mengurutkan secara ascending
unique_alamat4 = sorted(df_assessment['Alamat4'].unique())

# Menampilkan hasil
print(unique_alamat4)
```

```python
# Menampilkan seluruh nilai unik dari kolom Funcloc dalam bentuk list
unique_funcloc = df_assessment['Funcloc'].unique().tolist()

# Menampilkan hasil
print(unique_funcloc)
```

```python
# Menghitung jumlah nilai unik pada kolom Funcloc
num_unique_funcloc = len(df_assessment['Funcloc'].unique())

# Menampilkan hasil
print(num_unique_funcloc)
```

```python
# Menghitung panjang dari setiap nilai unik pada kolom Funcloc
funcloc_lengths = df_assessment['Funcloc'].astype(str).apply(len).unique()

# Menampilkan hasil
print(funcloc_lengths)
```

```python
# Mengonversi kolom Funcloc ke format string
df_assessment['Funcloc'] = df_assessment['Funcloc'].astype(str)
```

```python
df_assessment.info()
```

```python
# Menggunakan rank dengan method='dense' dan mengonversi hasil ke integer
df_assessment['Funcloc_Order'] = df_assessment['Funcloc'].rank(method='dense', ascending=True).astype(int)
```

```python
df_assessment.head()
```

```python
# Menggabungkan df_assessment dan df_city_indonesia menggunakan outer join
merged_df_assessment = pd.merge(df_assessment, df_city_indonesia, left_on='Alamat4', right_on='City', how='outer')
```

```python
merged_df_assessment['Sequence'] = merged_df_assessment.groupby('CityCode')['Funcloc'].rank(method='first').astype(int).astype(str).str.zfill(3)
```
