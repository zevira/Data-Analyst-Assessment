```python
import pandas as pd
import random
from datetime import datetime, timedelta
```

```python
# Function to generate unique ID
def generate_unique_id(existing_ids):
    while True:
        new_id = random.randint(1, 10000)
        if new_id not in existing_ids:
            existing_ids.add(new_id)
            return new_id

# Function to generate a random date between two given dates
def random_date(start_date, end_date):
    return start_date + timedelta(
        seconds=random.randint(0, int((end_date - start_date).total_seconds()))
    )
```

```python
# Generate branch_details data
branches = []
branch_codes = set()
for i in range(1, 501):
    branch_code = f"B{i:03}"
    branch_codes.add(branch_code)

    # Hanya ambil tanggal untuk operation_start_date dan last_renovation_date
    operation_start_date = random_date(datetime(2000, 1, 1), datetime(2024, 1, 1)).date()
    last_renovation_date = random_date(datetime(2000, 1, 1), datetime(2024, 1, 1)).date()

    branches.append({
        "branch_code": branch_code,
        "branch_name": f"Branch_{i}",
        "region_id": random.randint(1, 3),
        "operation_start_date": operation_start_date,
        "is_active": random.choice([1, 0]),
        "max_capacity": random.randint(100, 1000),
        "last_renovation_date": last_renovation_date
    })

branch_details = pd.DataFrame(branches)
```

```python
branch_details.head()
```

```python
# Generate staff_records data
staff = []
staff_ids = set()
for i in range(1, 501):
    staff_id = generate_unique_id(staff_ids)
    branch_code = random.choice(list(branch_codes))
    join_date = random_date(datetime(2015, 1, 1), datetime(2023, 1, 1))

    # Mengatur resignation_date jika random() > 0.7
    resignation_date = join_date + timedelta(days=random.randint(30, 2000)) if random.random() > 0.7 else None

    # Mengubah resignation_date menjadi tanggal (bukan datetime)
    if resignation_date is not None:
        resignation_date = resignation_date.date()
    else:
        resignation_date = '0'

    # Generate supervisor_id, pastikan tidak ada NaN
    supervisor_id = generate_unique_id(staff_ids)

    staff.append({
        "staff_id": staff_id,
        "branch_code": branch_code,
        "join_date": join_date.date(),
        "role_code": f"R{random.randint(1, 5):02}",
        "base_salary": round(random.uniform(20000, 80000), 2),
        "supervisor_id": supervisor_id,
        "resignation_date": resignation_date,
        "performance_score": round(random.uniform(1.0, 5.0), 2)
    })

staff_records = pd.DataFrame(staff)
```

```python
staff_records.head()
```

```python
# Generate product_hierarchy data
products = []
product_codes = set()
for i in range(1, 501):
    product_code = f"P{i:06}"
    product_codes.add(product_code)
    products.append({
        "product_code": product_code,
        "product_name": f"Product_{i}",
        "category_l1": random.choice(["Men's Clothing", "Women's Clothing", "Children's Clothing"]),
        "category_l2": random.choice(["Shirts", "Pants", "Jackets", "Sweaters", "T-Shirts"]),
        "category_l3": random.choice(["Casual", "Formal", "Sporty", "Party", "Outdoor"]),
        "unit_cost": round(random.uniform(10, 100), 2),
        "unit_price": round(random.uniform(20, 150), 2),
        "supplier_id": f"SUP{random.randint(1, 5):02}",
        "minimum_stock": random.randint(5, 50),
        "is_perishable": random.choice([1, 0])
    })
product_hierarchy = pd.DataFrame(products)
```

```python
product_hierarchy.head()
```

```python
# Generate inventory_movements data
movements = []
movement_ids = set()
for i in range(1, 501):
    movement_id = generate_unique_id(movement_ids)
    product_code = random.choice(list(product_codes))
    branch_code = random.choice(list(branch_codes))
    transaction_type = random.choice(["IN", "OUT", "ADJ", "RTN", "EXP"])

    # Hanya ambil tanggal untuk transaction_date
    transaction_date = random_date(datetime(2000, 1, 1), datetime(2024, 1, 1)).date()

    # Isi expiry_date untuk semua transaksi
    expiry_date = random_date(datetime(2000, 1, 1), datetime(2024, 1, 1)).date()

    movements.append({
        "movement_id": movement_id,
        "product_code": product_code,
        "branch_code": branch_code,
        "transaction_type": transaction_type,
        "quantity": random.randint(1, 1000),
        "transaction_date": transaction_date,
        "batch_number": f"BATCH{i:05}",
        "expiry_date": expiry_date,
        "unit_cost_at_time": round(random.uniform(10, 100), 2)
    })

inventory_movements = pd.DataFrame(movements)
```

```python
inventory_movements.head()
```

```python
# Generate sales_transactions data
sales = []
transaction_ids = set()
for i in range(1, 501):
    transaction_id = generate_unique_id(transaction_ids)
    branch_code = random.choice(list(branch_codes))
    staff_id = random.choice(list(staff_ids))

    # Ambil hanya tanggal untuk transaction_date
    transaction_date = random_date(datetime(2000, 1, 1), datetime(2024, 1, 1)).date()

    sales.append({
        "transaction_id": transaction_id,
        "branch_code": branch_code,
        "transaction_date": transaction_date,
        "staff_id": staff_id,
        "payment_method": random.choice(["Cash", "Card", "Online"]),
        "total_amount": round(random.uniform(50, 1000), 2),
        "discount_amount": round(random.uniform(0, 50), 2),
        "loyalty_points_earned": random.randint(0, 100),
        "loyalty_points_redeemed": random.randint(0, 50),
        "customer_id": f"CUST{random.randint(1, 100):05}"
    })

sales_transactions = pd.DataFrame(sales)
```

```python
sales_transactions.head()
```

```python
# Generate sales_line_items data
line_items = []
for i in range(1, 501):
    transaction_id = random.choice(list(transaction_ids))
    product_code = random.choice(list(product_codes))

    quantity = random.randint(1, 100)
    unit_price = round(random.uniform(500, 5000), 2)

    discount_percentage = round(random.uniform(0, 40), 2)
    total_line_amount = quantity * unit_price * (1 - discount_percentage / 100)

    line_items.append({
        "transaction_id": transaction_id,
        "line_number": i,
        "product_code": product_code,
        "quantity": quantity,
        "unit_price_at_time": unit_price,
        "discount_percentage": discount_percentage,
        "total_line_amount": round(total_line_amount, 2)
    })

sales_line_items = pd.DataFrame(line_items)

# Format pandas untuk menghindari scientific notation
pd.options.display.float_format = '{:,.2f}'.format
```

## Calculate inventory turnover metrics per branch

```python
import sqlite3
import pandas as pd

# Buat koneksi SQLite sementara
conn = sqlite3.connect(":memory:")

# Simpan data DataFrame ke SQLite
branch_details.to_sql("branch_details", conn, if_exists="replace", index=False)
inventory_movements.to_sql("inventory_movements", conn, if_exists="replace", index=False)

# Query SQL untuk menampilkan informasi yang diminta
query = """
WITH max_date AS (
    SELECT MAX(transaction_date) AS max_transaction_date
    FROM inventory_movements
),
stock_summary AS (
    SELECT
        bd.branch_name,
        im.branch_code,
        -- Total barang yang masuk (IN)
        SUM(CASE WHEN im.transaction_type = 'IN' THEN im.quantity ELSE 0 END) AS total_incoming_stock,
        -- Total barang yang keluar (OUT)
        SUM(CASE WHEN im.transaction_type = 'OUT' THEN im.quantity ELSE 0 END) AS total_outgoing_stock,
        -- Total penyesuaian stok (ADJ)
        SUM(CASE WHEN im.transaction_type = 'ADJ' THEN im.quantity ELSE 0 END) AS total_adjustment_stock,
        -- Total barang kadaluarsa (EXP)
        SUM(CASE WHEN im.transaction_type = 'EXP' THEN im.quantity ELSE 0 END) AS total_expired_stock,
        -- Total barang yang dikembalikan (RTN)
        SUM(CASE WHEN im.transaction_type = 'RTN' THEN im.quantity ELSE 0 END) AS total_returned_stock,
        -- Perhitungan stock level saat ini (penjumlahan dan pengurangan yang sesuai)
        SUM(CASE
            WHEN im.transaction_type = 'IN' THEN im.quantity  -- Barang masuk
            WHEN im.transaction_type = 'OUT' THEN -im.quantity  -- Barang keluar
            WHEN im.transaction_type = 'ADJ' THEN im.quantity  -- Penyesuaian stok
            WHEN im.transaction_type = 'RTN' THEN im.quantity  -- Retur barang (menambah stok)
            WHEN im.transaction_type = 'EXP' THEN -im.quantity  -- Barang kadaluarsa
            ELSE 0
        END) AS current_stock_level
    FROM
        inventory_movements im
    JOIN
        branch_details bd ON im.branch_code = bd.branch_code
    CROSS JOIN
        max_date
    WHERE
        im.transaction_date >= DATE(max_date.max_transaction_date, '-3 months')  -- Menggunakan max(transaction_date)
    GROUP BY
        bd.branch_name, im.branch_code
),
stock_turnover AS (
    SELECT
        branch_name,
        branch_code,
        total_incoming_stock,
        total_outgoing_stock,
        total_adjustment_stock,
        total_expired_stock,
        total_returned_stock,
        current_stock_level,
        -- Perhitungan stock turnover rate dengan rumus yang disesuaikan
        CASE
            WHEN total_incoming_stock = 0 THEN 0  -- Menghindari pembagian dengan nol
            ELSE CAST(total_outgoing_stock + total_adjustment_stock - total_expired_stock + total_returned_stock AS FLOAT) / total_incoming_stock
        END AS stock_turnover_rate
    FROM
        stock_summary
)
SELECT
    branch_name,
    total_incoming_stock,
    total_outgoing_stock,
    current_stock_level,
    stock_turnover_rate,
    total_expired_stock AS number_of_expired_items
FROM
    stock_turnover
ORDER BY
    stock_turnover_rate DESC;
"""

# Jalankan query dan simpan hasilnya ke DataFrame
result_inventory_turnover = pd.read_sql_query(query, conn)

pd.set_option("display.max_rows", None)

# Tampilkan hasil
print(result_inventory_turnover)

# Tutup koneksi SQLite
conn.close()
```

## Analyze staff performance and sales

```python
import sqlite3
import pandas as pd

# Create an in-memory SQLite database
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Creating the necessary tables (simplified for this example)
branch_details.to_sql("branch_details", conn, index=False, if_exists="replace")
staff_records.to_sql("staff_records", conn, index=False, if_exists="replace")
sales_transactions.to_sql("sales_transactions", conn, index=False, if_exists="replace")
sales_line_items.to_sql("sales_line_items", conn, index=False, if_exists="replace")

# SQL query to get the desired results
query = """
SELECT
    s.staff_id,
    b.branch_name,
    COUNT(st.transaction_id) AS number_of_transactions,
    SUM(st.total_amount) AS total_sales_amount,
    AVG(st.total_amount) AS average_transaction_value,
    s.performance_score
FROM
    staff_records s
JOIN
    branch_details b ON s.branch_code = b.branch_code
LEFT JOIN
    sales_transactions st ON s.staff_id = st.staff_id
WHERE
    s.resignation_date = '0'  -- Active staff (no resignation)
GROUP BY
    s.staff_id, b.branch_name, s.performance_score
"""

# Execute the query and load it into a DataFrame
result_staff_performance = pd.read_sql(query, conn)

# Close the connection
conn.close()
```

## Customer Purchase Patterns

```python
import sqlite3
import pandas as pd

# Create an in-memory SQLite database
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Create tables in SQLite using the previously generated DataFrames
sales_transactions.to_sql("sales_transactions", conn, index=False, if_exists="replace")
sales_line_items.to_sql("sales_line_items", conn, index=False, if_exists="replace")
product_hierarchy.to_sql("product_hierarchy", conn, index=False, if_exists="replace")

# SQL query to find products frequently bought together with frequency >= 10
query = """
SELECT
    sli1.product_code,
    st.transaction_date,
    COUNT(*) AS transaction_count
FROM
    sales_line_items sli1
JOIN
    sales_transactions st
    ON sli1.transaction_id = st.transaction_id  -- Join using transaction_id
GROUP BY
    sli1.product_code, st.transaction_date
HAVING
    COUNT(*) > 10;  -- Only consider products with more than 10 transactions on the same date
"""

# Execute the query and load it into a pandas DataFrame
result_customer_purchase_pattern = pd.read_sql(query, conn)

# Close the connection
conn.close()
```

## Data Visualization

```python
import pandas as pd

# Merge sales_line_items with sales_transactions to get transaction data
merged_sales = pd.merge(sales_line_items, sales_transactions, on='transaction_id', how='left')

# Total sales (revenue) per product
product_sales = merged_sales.groupby('product_code').agg(
    total_revenue=('total_line_amount', 'sum'),
    total_quantity=('quantity', 'sum')
).reset_index()

# Merge with product_hierarchy to get product names and categories
product_sales = pd.merge(product_sales, product_hierarchy[['product_code', 'product_name', 'category_l1', 'category_l2', 'category_l3']], on='product_code', how='left')
```

```python
product_sales
```

```python
# Mengekspor DataFrame ke File Excel
product_sales.to_excel('product_sales.xlsx', index=False)
```

```python
# Sales trends (by transaction date)
sales_trends = merged_sales.groupby('transaction_date').agg(
    total_daily_sales=('total_line_amount', 'sum'),
    total_daily_quantity=('quantity', 'sum')
).reset_index()
```
